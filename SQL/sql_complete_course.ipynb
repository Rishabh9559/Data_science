function_name(column) OVER (
    PARTITION BY column_name,
    ORDER BY column_name,
    ROWS or RANGE clause
)

```
* **PARTITION BY** - Divides the result set into logical group like **GROUP BY**. The function restart its calculation for each partition.

* **ORDER BY** - Sorts the rows within each partition

* **ROWS/RANGE** - Further restricts the window to a subset of rows relative to the current row.

* **ROW_NUMBER()** : Assigns a unique sequential number(1,2,3,4)

* **RANK()** : Gives the same number to ties but skips the next number (1,1,3,4)


* **DENSE_RANK()** : Gives the same number to ties but does not skip number (1,1,2,3)

```
SELECT Employee, Department, Revenue
ROW_NUMBER() OVER(PARTITION BY department ORDER BY revenue DESC) AS RowNum,
RANK() OVER(PARTITION BY department ORDER BY revenue DESC ) AS Rank,
DENSE_RANK() OVER(PARTITION BY Department ORDER BY Revenue DESC) AS denseRank
FROM Sales
```

### **LAG/LEAD**
 * LAG(column, offset) : Lokks back at previous rows

 * LEAD(column, offset) : Looks forward at following rows

```
