## **3.4 Operators in Python**
Operators are special symbols that carry out arithmetic, logical, and bitwise computations on operands (variables and values). Python categorizes operators into several types:



### 1. **Arithmetic Operators**

**Addition (+):** Adds two operands (e.g., a = 10, b = 10 -> a + b = 20).


**Subtraction (-):** Subtracts the second operand from the first (e.g., a = 20, b = 5 -> a - b = 15).


**Multiplication (*):** Multiplies two operands (e.g., a = 20, b = 4 -> a * b = 80).


**Division (/):** Divides the numerator by the denominator, returning a floating-point quotient (e.g., a = 20, b = 10 -> a / b = 2.0).


**Modulus (%):** Returns the division remainder (e.g., a = 20, b = 10 -> a % b = 0).


**Exponentiation ( ** ):** Raises the base operand to the power of the exponent (e.g., a = 2, b = 3 -> a ** b = 8).


**Floor Division (//):** Divides and rounds down to the nearest whole integer (e.g., a = 20, b = 3 -> a // b = 6).

### **5. Bitwise Operators**

**Bitwise AND (&):** Performs a bit-by-bit AND operation.


**Bitwise OR (|):** Performs a bit-by-bit OR operation.

**Bitwise XOR (^):** Performs a bit-by-bit Exclusive OR operation.

**Bitwise NOT / Complement (~):** Inverts all bits of the operand.

**Left Shift (<<):** Shifts binary bits to the left by the specified number of positions.

**Right Shift (>>):** Shifts binary bits to the right by the specified number of positions.


````python
# Initialize two variables
a = 12  # In binary: 1100
b = 10  # In binary: 1010

print(f"a = {a} (Binary: {bin(a)})")
print(f"b = {b} (Binary: {bin(b)})\n")

# 1. Bitwise AND (&)
# Compares each bit. Returns 1 if both bits are 1, otherwise 0.
# 1100 & 1010 = 1000 (which is 8 in decimal)
print("Bitwise AND (&):")
print("a & b =", a & b)

# 2. Bitwise OR (|)
# Compares each bit. Returns 1 if at least one bit is 1.
# 1100 | 1010 = 1110 (which is 14 in decimal)
print("\nBitwise OR (|):")
print("a | b =", a | b)

# 3. Bitwise XOR (^)
# Compares each bit. Returns 1 only if the bits are different.
# 1100 ^ 1010 = 0110 (which is 6 in decimal)
print("\nBitwise XOR (^):")
print("a ^ b =", a ^ b)

# 4. Bitwise NOT (~)
# Inverts all bits. In Python, ~x is equivalent to -(x + 1).
# ~12 = -(12 + 1) = -13
print("\nBitwise NOT (~):")
print("~a =", ~a)

# 5. Left Shift (<<)
# Shifts the bits of 'a' to the left by 2 positions. Adds zeros on the right.
# 1100 << 2 = 110000 (which is 48 in decimal)
print("\nLeft Shift (<<):")
print("a << 2 =", a << 2)

# 6. Right Shift (>>)
# Shifts the bits of 'a' to the right by 2 positions.
# 1100 >> 2 = 0011 (which is 3 in decimal)
print("\nRight Shift (>>):")
print("a >> 2 =", a >> 2)
````

````python
# sort the list value

lis = [10,5,7,1,8,2,6,]
print(f"Before sort list : {lis}")
for i in range(len(lis)):
  for j in range(i,len(lis)):
    if (lis[i]>lis[j]):
      temp = lis[i]
      lis[i]=lis[j]
      lis[j]=temp

print(f"After sort list : {lis}")

# Time complexity O(n^2)
````

## **6.2 While loops**
A while loop in Python repeatedly executes a block of code as long as a specified condition remains True. It is primarily used when the **total number of iterations is unknown** before the loop starts.




````python
# while loops

count = 5

while (count>1):
  print(count)
  count -=1
````

````python
# while loops

# Binary search
lists = [10,20,30,40,50] # Binary search- list must be in sorted order

target = 30

low = 0
high = len(lists)-1

# while loops
while(low<=high):
  mid = (low + high)//2
  if lists[mid]==target:
    print(f"found at index {mid}")
    break
  elif lists[mid]<target:
    low=mid+1
  else:
    high=mid-1
````

## **10.1 Encapsulation**

Encapsulation is the process of bundling **data(attributes)** and the **methods** that operate on that data into s **single unit** (class), while controlling how data can be **access and modified**.

-- Simple words: Encapsulation means wrapping data and methods together and restricting direct access to important data.


**Python does not have strict, compiler-enforced access modifiers. Instead, it relies on naming conventions (using underscores)**

**Types of Encapsulation**
* **Public** : Access from any parts of program, both outside and inside the program.
* **Protected** : Protected - Accessible only within the defining class and its subclasses (derived/child classes). It is hidden from unrelated outside code.
* **Private** : Accessible only within the exact class where they are declared. They cannot be seen or modified by outside code or derived classes


````python
class BankAccount:
  def __init__(self, account_holder, balance, pin):
    # public
    self.account_holder = account_holder

    # protected
    self._balance = balance

    # private
    self.__pin = pin

  def check_pin(self, user_input):
    return self.__pin == user_input

class SavingAccount(BankAccount):
  def addInterest(self, rate):
    interest = self._balance * rate # access the protected

    self._balance +=interest
    print(f"Interest added and updated amount {self._balance}")


my_account = SavingAccount("Rishabh", 500000, 1234)

# public access
print(f"Account holder : {my_account.account_holder}")

# Protected Access
my_account.addInterest(7.5/100)

# Private access
is_correct = my_account.check_pin(1234)
print(f"PIN verification {is_correct}")
````

## **10.2 Inheritance**

Inheritance is an OOP mechanism in which a child class acquires attributes and methods from a parent class.

**In simple words** : Inheritance allows us to reuse existing code from another class.

````python
# Single Inheritance

class Animal: # parents
  def eats(self):
    print("Animal eats foods")

class Dog(Animal): # child
  def bark(self):
    print("Dog barks")

#-------------
dog = Dog()
dog.bark()
dog.eats()
````

## **Types of Inheritance**

* Single Inheritance
* Multilevel Inheritance
* Multiple Inheritance
* Hierarchical Inheritance
* Hybrid Inheritance


### **Single Inheritance**

In Single Inheritance a derived (child) inherits froma single base (parent) class.

````python
# Single Inheritance
class Parent:
  def color(self):
    return "White"

# child class
class Child(Parent):
  def height(self):
    return "5.9 fits"

child = Child()
print(child.color())
print(child.height())
````

### **Multilevel Inheritance**

In multilevel Inheritance, a **child class** inherits from a **parent class**, which in turns inherits from another **grandparent class**, forming a liner chain

````python
# Multilevel inheritance

class GameCharacter:
  def __init__(self, name):
    self.name = name

  def introduce(self):
    print(f"I am {self.name} a game character")

  def move(self):
    print(f"{self.name} can move")


# child of GameCharacter
class Warrior(GameCharacter):
  def attack(self):
    print(f"{self.name} attacks with a sword")

# child of warrior
class SuperWarrior(Warrior):
  def specialPower(self):
    print(f"{self.name} uses super power")

  def fly(self):
    print(f"{self.name} can fly")


# creating objects

name = "Rishabh"
hero = SuperWarrior(name)

hero.introduce() # From GameCharacter
hero.move() # From GameCharacter
hero.attack() # From Warrior
hero.specialPower() # From superwarrior
hero.fly()  # From superwarrior
````

### **Multiple Inheritance**

Multiple Inheritance occurs when a **child** inherits directly from more than one parent class.

````python
# mutiple inheritance

class Camera:
  def takePhoto(self):
    print("Taking a high quality photo")

class MusicPlayer:
  def playMusic(self):
    print("Playing music")

# inheriting from two parent classess
class Smartphone(Camera, MusicPlayer):
  def makeCall(self):
    print("making a phone call")


# creating Object
phone = Smartphone() #Object
phone.takePhoto()  # From camera
phone.playMusic()   # from musicplayer
phone.makeCall()  # from smartphone
````

### **Hierarchical Inheritance**
Hierarchical Inheritance involves **mutile child class** inheriting from a **single shared parent class**

````python
#parent class
class Vehical:
  def start(self):
    print("Vehical is starting")

  def stop(self):
    print("Vehical is stopped")

# child class 1
class Car(Vehical):
  def drive(self):
    print("Car is driving")

# child class 2
class Bike(Vehical):
  def ride(self):
    print("Bike is riding")

# child class 3
class Truck(Vehical):
  def loadGoods(self):
    print("Truck is carrying people")

#creating object
car = Car()
bike = Bike()
truck = Truck()

car.start()
car.drive()

bike.start()
bike.ride()

truck.start()
truck.loadGoods()
````

### **Hybrid Inheritance**

Hybrid inheritance is a combination of **two or more inheritance** types combining multiple and Hierarchical inheritance

````python
# Parent class
class Person:
  def showPerson(self):
    print("I am a person")

# child class 1
class Student(Person):
  def study(self):
    print("I am studying Computer Science")

# child class 2
class Employee(Person):
  def work(self):
    print("I am working as a Software developer")

# Mutiple Inheritance
class TeachingAssistant(Student, Employee):
  def assist(self):
    print("I help the professor and guide students")

# creating object
teach = TeachingAssistant()

teach.showPerson()
teach.study()
teach.work()
teach.assist()
````

# **10.3 Polymorphism**

Polymorphism means **many forms**.

Its allows single function, method and object interface to work with different data types.

Polymorphism allows the same method name, function, or operator to behave differently depending on the object or data type.

Types of Polymorphism:

1. **Method Overriding**
2. **Method Overloading**

**1. Method Polymorphism**

Method overriding occure when a child class define a method with same name as a method in its parent class.

It is also called **Runtime polymorphism**

````python
class Animal:
  def sound(self):
    print("Animal make a sound")

class Dog (Animal):
  def sound(self):
    print("Dog barks")

class Cat (Animal):
  def sound(self):
    print("Cat meows")

#object
a = Animal()
d = Dog()
c = Cat()

a.sound()
d.sound()
c.sound()
````

**2. Method Overloading**

Method overloading traditionally means defining multiple method with the same name but different parameter list.

Pythod **does not support** traditional method overloading by defining multiple methods with the same name in one class.

We can achieve similar behiver with default arguments

````python
class Calculator:
  def add(self, a, b=0, c=0):
    return a + b + c

# object
calc = Calculator()
print(calc.add(10))
print(calc.add(10,20))
print(calc.add(10,20,40))
````

**3. Operator Overloading**

Operator overloading allows operators to works with object of user define classes.


````python
# The + operator can be customized using special method __add__()

class student:
  def __init__(self, marks):
    self.marks = marks

  def __add__(self,other):
    return self.marks * other.marks

# object
s1 = student(10)
s2 = student(10)

print(s1+s2)
````

# **10.4 Abstraction**

Abstraction means **hiding** internal implementation details and showing only the essential feature to the user.



**Real-life example**

Consider an ATM machine. You can withdraw money by entering your PIN and amount, but you do not need to know how the bank verifies your account or processes the transaction internally.
- What you see: Enter PIN, select amount, withdraw money.
- What is hidden: Account verification, database operations, and transaction processing.

Types of Abstraction
1. **Data Abstraction** : Hides internal data and implementation details while exposing only the necessary information or operations through an interface.

2. **Abstract class and Method based abstraction** : Use python **abc** module to define methods that subclasses must implement.

````python
# abc ->  Abstract Base Classes
from abc import ABC, abstractmethod

class Animal(ABC):
  @abstractmethod
  def sound(self):
    pass

class Dog(Animal):
  def sound(self):
    print("Dog barks")

class Cat(Animal):
  def sound(self):
    print("cat meows")

# object
dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
````
